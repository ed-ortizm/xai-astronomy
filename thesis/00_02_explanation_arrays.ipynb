# XAI array

````python
bin_id = "bin_03"
xai_run_ids = glob.glob(f"{explanations_dir}/{bin_id}/*/xai*")

for idx, xai_run_id_path in enumerate(xai_run_ids):

    score_name = xai_run_id_path.split("/")[-2]
    run_id = xai_run_id_path.split("/")[-1]

    if '100000' in run_id:
        print(run_id)
    
    specobjids = np.load(
        f"{explanations_dir}/{bin_id}/{score_name}/specobjids_anomalies.npy"
    )

    explanation_array = ingest_explanations_into_array(
        specobjids_array=specobjids,
        wave=wave,
        load_from_dir=xai_run_id_path,
    )

    print(f"{idx}\t{score_name}:\t{run_id}")
    np.save(
        f"{explanations_dir}/{bin_id}/{score_name}/explanation_array_{run_id}.npy",
        explanation_array,
    )
````
